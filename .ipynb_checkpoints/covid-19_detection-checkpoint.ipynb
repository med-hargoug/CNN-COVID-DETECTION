# redimensionnées les images  

```python
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
import seaborn as sns
from sklearn.metrics import classification_report

IMG_SIZE = (128, 128)

def load_images(df, img_dir, img_size):
    images = []
    labels = []
    for _, row in df.iterrows():
        img_path = os.path.join(img_dir, row['filename'])
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        if img is not None:
            img = cv2.resize(img, img_size)
            images.append(img)
            labels.append(row['label'])
    return np.array(images), np.array(labels)

print("Loading images...")
X, y = load_images(df, IMG_DIR, IMG_SIZE)
print(f"Loaded {len(X)} images — shape: {X.shape}")
```

# Normalisation des donner et préparation des donner de entrainement et de teste  :

```python
# Normalize (same idea as the MNIST lab: divide by 255)
X_scaled = X / 255.0

# Stratified split so both classes are represented in train & test
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Train: {X_train.shape} | Test: {X_test.shape}")
print(f"Train — Non-COVID: {(y_train==0).sum()}, COVID: {(y_train==1).sum()}")
print(f"Test  — Non-COVID: {(y_test==0).sum()},  COVID: {(y_test==1).sum()}")
```

# Equilibrage de Target :

```python
# Class weights — compensate for 3073 vs 406 imbalance
class_weights_array = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_train),
    y=y_train
)
class_weights = dict(enumerate(class_weights_array))
print("Class weights:", class_weights)
```

# ANN :
    model build 

```python
ann = keras.Sequential([
    keras.layers.Input((128, 128)),
    keras.layers.Flatten(),
    keras.layers.Dense(256, activation='relu'),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])
ann.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

ann.summary()
```

# Training 

```python
history1 = ann.fit(
    X_train, y_train,
    epochs=20,
    validation_split=0.3,
    batch_size=32,
    class_weight=class_weights
)
```

```python
figure, axis = plt.subplots(ncols=2, figsize=(15, 5))
axis[0].plot(history1.history['loss'])
axis[0].plot(history1.history['val_loss'])
axis[0].set_xlabel('Epochs')
axis[0].set_ylabel('Loss')
axis[0].legend(['train', 'test'])
axis[1].plot(history1.history['accuracy'])
axis[1].plot(history1.history['val_accuracy'])
axis[1].set_xlabel('Epochs')
axis[1].set_ylabel('Accuracy')
axis[1].legend(['train', 'test'])
plt.show()
```

# Evaluation

```python
train_evaluation = ann.evaluate(X_train, y_train)
test_evaluation  = ann.evaluate(X_test,  y_test)
print("-" * 80)
print(f"Train Loss = {train_evaluation[0]}, Train Accuracy = {train_evaluation[1]}")
print("-" * 80)
print(f"Test  Loss = {test_evaluation[0]},  Test  Accuracy = {test_evaluation[1]}")
print("-" * 80)
```

```python
pred = (ann.predict(X_test) > 0.5).astype(int).flatten()
print(classification_report(y_test, pred, target_names=['Non-COVID', 'COVID']))
```

```python
cm = tf.math.confusion_matrix(labels=y_test, predictions=pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds',
            xticklabels=['Non-COVID', 'COVID'],
            yticklabels=['Non-COVID', 'COVID'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

## CNN 
    model build

```python
# CNN needs an extra channel dimension: (N, 128, 128, 1)
X_train_cnn = X_train[..., np.newaxis]
X_test_cnn  = X_test[..., np.newaxis]
```

```python
cnn = keras.Sequential([
    keras.layers.Input((128, 128, 1)),
    keras.layers.Conv2D(filters=32, kernel_size=(3, 3), activation='relu'),
    keras.layers.MaxPool2D((2, 2)),
    keras.layers.Conv2D(filters=64, kernel_size=(3, 3), activation='relu'),
    keras.layers.MaxPool2D((2, 2)),
    keras.layers.Conv2D(filters=128, kernel_size=(3, 3), activation='relu'),
    keras.layers.MaxPool2D((2, 2)),
    keras.layers.Flatten(),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(256, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])
cnn.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

cnn.summary()
```

# training 

```python
history3 = cnn.fit(
    X_train_cnn, y_train,
    epochs=20,
    validation_split=0.3,
    batch_size=32,
    class_weight=class_weights
)
```

```python
figure, axis = plt.subplots(ncols=2, figsize=(15, 5))
axis[0].plot(history3.history['loss'])
axis[0].plot(history3.history['val_loss'])
axis[0].set_xlabel('Epochs')
axis[0].set_ylabel('Loss')
axis[0].legend(['train', 'test'])
axis[1].plot(history3.history['accuracy'])
axis[1].plot(history3.history['val_accuracy'])
axis[1].set_xlabel('Epochs')
axis[1].set_ylabel('Accuracy')
axis[1].legend(['train', 'test'])
plt.show()
```

# Evaluation

```python
train_evaluation = cnn.evaluate(X_train_cnn, y_train)
test_evaluation  = cnn.evaluate(X_test_cnn,  y_test)
print("-" * 80)
print(f"Train Loss = {train_evaluation[0]}, Train Accuracy = {train_evaluation[1]}")
print("-" * 80)
print(f"Test  Loss = {test_evaluation[0]},  Test  Accuracy = {test_evaluation[1]}")
print("-" * 80)
```

```python
pred = (cnn.predict(X_test_cnn) > 0.5).astype(int).flatten()
print(classification_report(y_test, pred, target_names=['Non-COVID', 'COVID']))
```

```python
cm = tf.math.confusion_matrix(labels=y_test, predictions=pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds',
            xticklabels=['Non-COVID', 'COVID'],
            yticklabels=['Non-COVID', 'COVID'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
img_resized = tf.image.resize(img[..., np.newaxis], size=(128, 128))
img_resized.shape
```

```python
# CNN prediction
predicted = cnn.predict(np.expand_dims(img_resized / 255, 0))
label = 'COVID' if predicted[0][0] > 0.5 else 'Non-COVID'
confidence = predicted[0][0] if predicted[0][0] > 0.5 else 1 - predicted[0][0]
print(f"CNN predicts: {label}  (confidence: {confidence:.2%})")
```

```python
# ANN prediction
img_flat = tf.image.resize(img[..., np.newaxis], size=(128, 128))[:, :, 0]
predicted_ann = ann.predict(np.expand_dims(img_flat / 255, 0))
label_ann = 'COVID' if predicted_ann[0][0] > 0.5 else 'Non-COVID'
confidence_ann = predicted_ann[0][0] if predicted_ann[0][0] > 0.5 else 1 - predicted_ann[0][0]
print(f"ANN predicts: {label_ann}  (confidence: {confidence_ann:.2%})")
```
